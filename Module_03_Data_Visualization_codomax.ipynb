```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

1. Load dataset

```python
df = pd.read_csv("student_data.csv")
df
```

2. handle missing values

```python
df.fillna(df['Math'].mean(), inplace=True)
df
```

3. bar chart

```python
df.columns = df.columns.str.strip()

plt.figure(figsize=(8,5))
plt.bar(df["Name"], df["Math"], color="skyblue")
plt.title("Math Scores of Students")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

Insight: Diya scored the highest marks in Mathematics.

4. line chart

```python
plt.figure(figsize=(8,5))
plt.plot(df["Name"], df["Science"], marker="o", color="green")
plt.title("Science Scores Trend")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

Insight: Science scores remain comparatively consistent among students.

5. pie chart


```python
subject_totals = [
    df["Math"].sum(),
    df["Science"].sum(),
    df["English"].sum()
]

subjects = ["Math", "Science", "English"]

plt.figure(figsize=(6,6))
plt.pie(subject_totals,
        labels=subjects,
        autopct="%1.1f%%")
plt.title("Subject-wise Score Distribution")
plt.show()
```

Insight: English contributes the highest share of overall marks.

6. Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df["Math"],
         bins=5,
         color="orange",
         edgecolor="black")
plt.title("Distribution of Math Scores")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

Insight: Most students score between 75 and 90 marks.

7. Scatter plot

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=df["Math"],
    y=df["Science"],
    s=100
)

plt.title("Math vs Science Scores")
plt.xlabel("Math")
plt.ylabel("Science")
plt.show()
```

Insight: Students performing well in Mathematics generally perform well in Science too.
